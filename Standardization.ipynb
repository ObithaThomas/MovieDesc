```python
import cv2
import os

# Input and output paths
input_dir = r"downloaded_videos"
output_dir = r"standardized_videos"

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Standard video properties
FRAME_WIDTH = 640  # Resize width
FRAME_HEIGHT = 480  # Resize height
FPS = 30  # Standard FPS

# Process each video in the input directory
for filename in os.listdir(input_dir):
    if filename.endswith(".mp4"):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        # Open video file
        cap = cv2.VideoCapture(input_path)
        if not cap.isOpened():
            print(f"❌ Error opening {filename}")
            continue
        
        # Get original properties
        original_fps = int(cap.get(cv2.CAP_PROP_FPS)) or FPS
        fourcc = cv2.VideoWriter_fourcc(*"mp4v")  # Codec

        # Create video writer
        out = cv2.VideoWriter(output_path, fourcc, FPS, (FRAME_WIDTH, FRAME_HEIGHT))

        print(f"🔄 Standardizing {filename}...")

        while True:
            ret, frame = cap.read()
            if not ret:
                break  # End of video

            # Resize frame
            frame = cv2.resize(frame, (FRAME_WIDTH, FRAME_HEIGHT))

            # Write standardized frame
            out.write(frame)

        # Release resources
        cap.release()
        out.release()
        print(f"✅ Standardized: {filename}")

print("🎉 Standardization complete for all videos!")
```

```python
import cv2
import os
import psutil
import time

# Input and output paths
input_dir = r"downloaded_videos"
output_dir = r"standardized_videos"
# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)
# Standard video properties
FRAME_WIDTH = 640  # Standardized width
FRAME_HEIGHT = 480  # Standardized height
STANDARD_FPS = 30  # Standard FPS
# List video files in the input directory
video_files = [f for f in os.listdir(input_dir) if f.endswith(".mp4")]
# To store the lowest resolution and FPS from all videos
lowest_resolution = (float('inf'), float('inf'))
lowest_fps = float('inf')
# Analyze all video files to find the lowest resolution and FPS
for filename in video_files:
    input_path = os.path.join(input_dir, filename)
    cap = cv2.VideoCapture(input_path)
    if not cap.isOpened():
        print(f"❌ Error opening {filename}")
        continue
    # Get the video properties
    original_fps = int(cap.get(cv2.CAP_PROP_FPS)) or STANDARD_FPS  # Default to STANDARD_FPS if it's 0
    original_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    original_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    # Track the lowest resolution and FPS
    lowest_resolution = min(lowest_resolution, (original_width, original_height))
    lowest_fps = min(lowest_fps, original_fps)
    cap.release()
# Output the lowest resolution and FPS
print(f"Lowest Resolution: {lowest_resolution[0]}x{lowest_resolution[1]}")
print(f"Lowest FPS: {lowest_fps}")

# Track memory usage before standardization
process = psutil.Process(os.getpid())
memory_before = process.memory_info().rss / (1024 * 1024)  # in MB
print(f"Memory usage before standardization: {memory_before:.2f} MB")

# Standardize all video files based on lowest resolution and FPS
for filename in video_files:
    input_path = os.path.join(input_dir, filename)
    output_path = os.path.join(output_dir, filename)
    
    cap = cv2.VideoCapture(input_path)
    if not cap.isOpened():
        print(f"❌ Error opening {filename}")
        continue
    
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")  # Codec for output
    out = cv2.VideoWriter(output_path, fourcc, lowest_fps, lowest_resolution)

    print(f"🔄 Standardizing {filename}...")

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Resize frame to lowest resolution
        frame = cv2.resize(frame, lowest_resolution)
        
        # Write the standardized frame
        out.write(frame)

    cap.release()
    out.release()

    print(f"✅ Standardized: {filename}")

# Track memory usage after standardization
memory_after = process.memory_info().rss / (1024 * 1024)  # in MB
print(f"Memory usage after standardization: {memory_after:.2f} MB")

# Calculate memory difference
memory_diff = memory_before - memory_after
print(f"Memory decrement: {memory_diff:.2f} MB")

print("🎉 Standardization complete for all videos!")
```
